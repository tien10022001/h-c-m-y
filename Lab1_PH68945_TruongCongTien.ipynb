## Bài 1: Làm quen với Numpy và Pandas

## Bài 3: Chạy thử mô hình đơn giản với Scikit-Learn
- Sử dụng tập dữ liệu có sẵn `iris` từ `sklearn.datasets`.
- Khởi tạo mô hình đơn giản `LinearRegression`.
- Thực hiện các bước: Load dữ liệu -> Khởi tạo model -> Fit model (huấn luyện) -> Predict (dự đoán).

```python
from sklearn.datasets import load_iris
from sklearn.linear_model import LinearRegression

# Bước 1: Load dữ liệu iris
iris = load_iris()
X = iris.data
y = iris.target
print(f'1. Kích thước tập Iris: X = {X.shape}, y = {y.shape}')

# Bước 2: Khởi tạo mô hình Linear Regression
model = LinearRegression()

# Bước 3: Fit model (huấn luyện)
model.fit(X, y)
print('2. Đã huấn luyện mô hình Linear Regression thành công.')

# Bước 4: Predict (dự đoán thử trên 5 mẫu đầu tiên)
y_pred = model.predict(X[:5])
print('\n3. Dự đoán 5 mẫu đầu tiên:')
print('- Nhãn thực tế : ', y[:5])
print('- Giá trị dự đoán:', np.round(y_pred, 2))
```

## Bài 4: Quy trình ML hoàn chỉnh - Phần 1: Thu thập và Kiểm tra dữ liệu
- Tải tập dữ liệu `digits` từ sklearn.
- In ra hình dạng (shape) của dữ liệu và nhãn.
- Hiển thị một mẫu dữ liệu dưới dạng hình ảnh.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits

# 1. Tải tập dữ liệu digits từ sklearn
digits = load_digits()

# 2. In ra hình dạng (shape) của dữ liệu và nhãn
print(f'1. Shape của dữ liệu X (features): {digits.data.shape}')
print(f'2. Shape của nhãn y (target): {digits.target.shape}')
print(f'3. Shape ma trận hình ảnh (images): {digits.images.shape}')

# 3. In giá trị của mẫu đầu tiên
print(f'\n4. Nhãn của mẫu đầu tiên: {digits.target[0]}')
print('Ma trận giá trị điểm ảnh 8x8 của mẫu đầu tiên:')
print(digits.images[0])

# 4. Hiển thị một mẫu dữ liệu dưới dạng hình ảnh
plt.figure(figsize=(3, 3))
plt.imshow(digits.images[0], cmap='gray')
plt.title(f'Mẫu đầu tiên (Nhãn: {digits.target[0]})')
plt.axis('off')
plt.show()
```
