## Bài 1: Làm sạch dữ liệu
**Yêu cầu:**
- Tạo một DataFrame chứa dữ liệu bị thiếu (NaN) và có giá trị ngoại lệ (outlier).
- Sử dụng Pandas để:
  1. Điền các giá trị thiếu bằng giá trị trung bình (mean) của cột.
  2. Phát hiện và xử lý giá trị ngoại lệ đơn giản (ví dụ: các giá trị lớn hơn 3 lần độ lệch chuẩn).

```python
import numpy as np
import pandas as pd

# Tạo DataFrame chứa dữ liệu bị thiếu (NaN) và có giá trị ngoại lệ (outlier)
data_b1 = {
    'Tuoi': [22, 25, np.nan, 23, 24, 26, np.nan, 25, 23, 22, 24, 25, 23, 26, 24],
    'ThuNhap': [15, 18, 16, 17, 150, 19, 17, np.nan, 18, 16, 17, 18, 19, 16, 15]  # 150 là outlier
}
df = pd.DataFrame(data_b1)
print("DataFrame ban đầu:")
print(df)

# 1. Điền các giá trị thiếu bằng giá trị trung bình (mean) của cột
df_filled = df.copy()
df_filled['Tuoi'] = df_filled['Tuoi'].fillna(df_filled['Tuoi'].mean())
df_filled['ThuNhap'] = df_filled['ThuNhap'].fillna(df_filled['ThuNhap'].mean())
print("\n1. DataFrame sau khi điền giá trị thiếu bằng mean:")
print(df_filled)

# 2. Phát hiện và xử lý giá trị ngoại lệ đơn giản (lớn hơn 3 lần độ lệch chuẩn)
mean_thunhap = df_filled['ThuNhap'].mean()
std_thunhap = df_filled['ThuNhap'].std()
upper_limit = mean_thunhap + 3 * std_thunhap
lower_limit = mean_thunhap - 3 * std_thunhap

print(f"\nMean ThuNhap: {mean_thunhap:.2f}, Std ThuNhap: {std_thunhap:.2f}")
print(f"Khoảng hợp lệ: [{lower_limit:.2f}, {upper_limit:.2f}]")

# Phát hiện các dòng ngoại lệ
outliers = df_filled[(df_filled['ThuNhap'] > upper_limit) | (df_filled['ThuNhap'] < lower_limit)]
print("\nCác dòng chứa giá trị ngoại lệ:")
print(outliers)

# Xử lý ngoại lệ: loại bỏ các dòng chứa giá trị ngoại lệ
df_cleaned = df_filled[(df_filled['ThuNhap'] <= upper_limit) & (df_filled['ThuNhap'] >= lower_limit)]
print("\n2. DataFrame sau khi loại bỏ ngoại lệ:")
print(df_cleaned)
```

## Bài 2: Chuẩn hóa dữ liệu (Scaling)
**Yêu cầu:**
- Tạo một mảng dữ liệu có các cột với đơn vị khác nhau (ví dụ: Tuổi từ 20-60 và Thu nhập từ 10tr-100tr).
- Sử dụng StandardScaler và MinMaxScaler từ thư viện sklearn.preprocessing để chuẩn hóa mảng dữ liệu này.
- In kết quả sau khi chuẩn hóa để so sánh.

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# Tạo mảng dữ liệu có các cột với đơn vị khác nhau
data_b2 = np.array([
    [22, 12000000],
    [28, 25000000],
    [35, 45000000],
    [42, 60000000],
    [50, 85000000],
    [58, 98000000]
])

print("Mảng dữ liệu gốc (Tuổi, Thu nhập):")
print(data_b2)

# Chuẩn hóa bằng StandardScaler
scaler_standard = StandardScaler()
data_standard = scaler_standard.fit_transform(data_b2)
print("\nKết quả chuẩn hóa bằng StandardScaler (Mean=0, Std=1):")
print(data_standard)

# Chuẩn hóa bằng MinMaxScaler
scaler_minmax = MinMaxScaler()
data_minmax = scaler_minmax.fit_transform(data_b2)
print("\nKết quả chuẩn hóa bằng MinMaxScaler ([0, 1]):")
print(data_minmax)

print("\nSo sánh:")
print("- StandardScaler đưa phân phối về chuẩn với mean = 0 và std = 1, phù hợp cho các thuật toán giả định phân phối chuẩn.")
print("- MinMaxScaler đưa dữ liệu về khoảng [0, 1], giữ nguyên tỉ lệ ban đầu, phù hợp khi dữ liệu không có phân phối Gauss hoặc có giới hạn cố định.")
```

## Bài 3: Chia tập dữ liệu Train/Test
**Yêu cầu:**
- Sử dụng tập dữ liệu diabetes từ sklearn.datasets.
- Sử dụng hàm train_test_split để chia dữ liệu thành 2 tập: train (80%) và test (20%).
- In ra kích thước của tập dữ liệu gốc và kích thước của hai tập sau khi chia.

```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split

# Tải tập dữ liệu diabetes
diabetes = load_diabetes()
X_diabetes = diabetes.data
y_diabetes = diabetes.target

# Chia dữ liệu: train (80%) và test (20%)
X_train_d, X_test_d, y_train_d, y_test_d = train_test_split(
    X_diabetes, y_diabetes, test_size=0.2, random_state=42
)

# In kích thước
print("Kích thước tập dữ liệu gốc:")
print(f"X: {X_diabetes.shape}, y: {y_diabetes.shape}")
print(f"Tổng số mẫu: {X_diabetes.shape[0]}, Số đặc trưng: {X_diabetes.shape[1]}")

print("\nKích thước sau khi chia:")
print(f"Tập huấn luyện (Train 80%): X_train: {X_train_d.shape}, y_train: {y_train_d.shape}")
print(f"Tập kiểm tra (Test 20%):    X_test:  {X_test_d.shape}, y_test:  {y_test_d.shape}")
```

## Bài 4: Quy trình ML hoàn chỉnh - Phần 2: Tiền xử lý dữ liệu
**Yêu cầu:**
- Tiếp tục với tập dữ liệu digits từ Lab 1.
- Thực hiện các bước tiền xử lý:
  1. Chuẩn hóa dữ liệu (Scaling) vì giá trị pixel nằm trong khoảng 0-16.
  2. Chia tập dữ liệu thành Train và Test.
  3. Xây dựng một Pipeline đơn giản chứa bước chuẩn hóa và một model (ví dụ: SVM hoặc Logistic Regression).

```python
from sklearn.datasets import load_digits
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Tải tập dữ liệu digits từ Lab 1
digits = load_digits()
X_digits = digits.data
y_digits = digits.target

print(f"Kích thước dữ liệu digits: {X_digits.shape}")
print(f"Khoảng giá trị pixel ban đầu: Min = {X_digits.min()}, Max = {X_digits.max()}")

# 1. Chuẩn hóa dữ liệu (Scaling)
scaler_digits = MinMaxScaler()
X_digits_scaled = scaler_digits.fit_transform(X_digits)
print(f"\n1. Sau khi chuẩn hóa (MinMaxScaler): Min = {X_digits_scaled.min()}, Max = {X_digits_scaled.max()}")

# 2. Chia tập dữ liệu thành Train (80%) và Test (20%)
X_train, X_test, y_train, y_test = train_test_split(
    X_digits, y_digits, test_size=0.2, random_state=42
)
print(f"\n2. Chia tập dữ liệu:")
print(f"- Kích thước tập Train: {X_train.shape}, {y_train.shape}")
print(f"- Kích thước tập Test:  {X_test.shape}, {y_test.shape}")

# 3. Xây dựng một Pipeline đơn giản chứa bước chuẩn hóa và một model (SVM)
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('model', SVC(kernel='rbf', C=1.0))
])

# Huấn luyện Pipeline trên tập Train
pipeline.fit(X_train, y_train)

# Đánh giá kết quả trên tập Test
y_pred = pipeline.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print("\n3. Pipeline: StandardScaler -> SVC")
print(f"Độ chính xác (Accuracy) trên tập Test: {accuracy * 100:.2f}%")
```
